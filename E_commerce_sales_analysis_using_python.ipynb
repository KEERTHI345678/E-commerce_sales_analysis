Task 11: Outlier Treatment Task 11: Outlier Treatment

- Identify outliers in the TotalAmount column using boxplot
- Calculate the IQR and remove outliers beyond 1.5×IQ
- Compare statistics before and after outlier removal
- Implement winsorization (capping) at 5th and 95th percentiles

```python
sns.boxplot(Merged_data['Total_Amount'])
q1=Merged_data['Total_Amount'].quantile(0.25)
q3=Merged_data['Total_Amount'].quantile(0.75)
IQR=q3=q1
lb=q1-1.5*IQR
ub=q3+1.5*IQR
print(lb)
print(ub)
outlier=Merged_data[(Merged_data['Total_Amount']<ub) &(Merged_data['Total_Amount']>lb)]
outlier=outlier['Total_Amount'].clip(upper=ub,lower=lb)
```

```python
print(f"The statistics before removing outliers: {Merged_data['Total_Amount'].describe()}")
print("***********************************************************************************")
print(f"The statistics after removing outliers: {outlier.describe()}")
```

```python
low=Merged_data['Total_Amount'].quantile(0.05)
up=Merged_data['Total_Amount'].quantile(0.95)
print(low)
print(up)
print("Before capping")
print(Merged_data['Total_Amount'].describe())
print("**************************************************************")
print("After capping")
Merged_data['Total_sales_cap']=Merged_data["Total_Amount"].clip(upper=up,lower=low)
print(Merged_data['Total_sales_cap'].describe())
```
